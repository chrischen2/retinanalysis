```python
# Single-cell (patch) experiments are stored with is_mea=0. Show a scrollable
# table of just the identifying / provenance columns.
# NOTE: experimenter / project / rig are not populated in the DB for patch data
# (only exp_name, species and rig_type carry values). species comes from the
# animal JSON (attributes.label); rig_type is the populated "rig" info ('PATCH').
import json
from IPython.display import display, HTML

sc_exps = ra.schema.Experiment() & 'is_mea=0'
df = sc_exps.to_pandas()
df['species'] = df['attributes'].apply(
    lambda d: (d if isinstance(d, dict) else json.loads(d)).get('label'))

cols = ['exp_name', 'species', 'experimenter', 'project', 'rig', 'rig_type']
df_view = df[cols].sort_values('exp_name').reset_index(drop=True)

print(f'Found {len(df_view)} single-cell experiments.')
# Scrollable, fixed-height box so every row is reachable without truncation.
display(HTML(
    '<div style="max-height:400px; overflow:auto;">'
    + df_view.to_html(index=False, max_rows=None)
    + '</div>'
))
```

```python
# Find single-cell blocks that ran a given protocol (case-insensitive substring match).
# Replaces the removed ra.get_datasets_from_protocol_names_sc helper; MEA discovery
# still lives in ra.get_datasets_from_protocol_names.
import pandas as pd

def get_sc_blocks_for_protocol(protocol_search: str) -> pd.DataFrame:
    prot = ra.schema.Protocol() & f'name LIKE "%{protocol_search}%"'
    blocks = ra.schema.EpochBlock() & [f'protocol_id={p}' for p in prot.fetch('protocol_id')]
    sc_ids = set(int(x) for x in (ra.schema.Experiment() & 'is_mea=0').fetch('id'))
    exp_map = {int(i): n for i, n in zip(*ra.schema.Experiment().fetch('id', 'exp_name'))}
    prot_map = {int(i): n for i, n in zip(*prot.fetch('protocol_id', 'name'))}
    rows = [
        {'exp_name': exp_map[int(eid)], 'protocol_name': prot_map[int(pid)], 'block_id': int(bid)}
        for bid, eid, pid in zip(*blocks.fetch('id', 'experiment_id', 'protocol_id'))
        if int(eid) in sc_ids
    ]
    df = pd.DataFrame(rows).sort_values(['exp_name', 'block_id']).reset_index(drop=True)
    print(f"Found {df.shape[0]} single-cell blocks across {df['exp_name'].nunique()} "
          f"experiments for {protocol_search!r}.")
    return df

get_sc_blocks_for_protocol('variableMeanNoise')
```

For Cell1 (OFF-parasol) let's get the expanding spots data. We pick a **cell-attached** block on purpose — we're counting spikes below, and whole-cell recordings have no extracellular spikes to detect.

```python
df_exp = ra.get_exp_summary(exp_name)
cell_label = 'Cell1'
df_exp.query('cell_label == @cell_label')
```

```python
# Grab a cell-attached ExpandingSpots block for this cell dynamically rather than
# hardcoding a block_id -- block IDs are DB auto-increments and shift whenever the
# database is repopulated, so a pinned integer goes stale. We require cell-attached
# (spiking) since the analysis below counts spikes.
df_es = df_exp[df_exp['cell_label'].eq(cell_label)
               & df_exp['protocol_name'].str.contains('ExpandingSpots', na=False)
               & df_exp['recording_technique'].eq('cell-attached')]
# Use the longest such block (most epochs / spot sizes).
block_id = int(df_es.sort_values('duration_minutes').iloc[-1]['block_id'])
print(f'Using block_id {block_id} for {exp_name} {cell_label} ExpandingSpots (cell-attached)')

sb = ra.StimBlock(exp_name, block_id)

# Note we're using SCResponseBlock. b_spiking=True will run spike detection.
rb = ra.SCResponseBlock(exp_name, block_id, b_spiking=True)
```

```python
# Example trace from 3rd epoch
def plot_eg_trace(rb, sb, i_epoch):
    f, ax = plt.subplots(1, 1, figsize=(10, 5))
    raw = rb.amp_data[i_epoch]
    sts = rb.spike_times[i_epoch]
    time = np.arange(len(raw)) / rb.amp_sample_rate
    ax.plot(time, raw, label='Raw trace')
    ax.scatter(sts/rb.amp_sample_rate, raw[sts], color='r')
    pre_time = sb.df_epochs.at[i_epoch, 'preTime']
    # Hacky correction for first missing frame. Need to implement frame monitor based timing.
    pre_time -= 1/60 * 1000
    stim_time = sb.df_epochs.at[i_epoch, 'stimTime']
    offset_time = pre_time + stim_time
    ax.axvline(pre_time / 1000, c='k', label='Stim onset')
    ax.axvline(offset_time / 1000, c='grey', label='Stim offset')
    spot_size = sb.df_epochs.at[i_epoch, 'currentSpotSize']
    ax.set_title(f'Epoch index {i_epoch}\n{spot_size}um spot')
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Amplitude')
    ax.legend()

i_epoch = 2
plot_eg_trace(rb, sb, i_epoch)
```

```python
def get_stim_nspikes(sb, rb, b_sub_baseline: bool=False):
    ls_nsps = []
    for e_idx in sb.df_epochs.index:
        pre_time = sb.df_epochs.at[e_idx, 'preTime']
        
        # Hacky correction for first missing frame. Need to implement frame monitor based timing.
        pre_time -= 1/60 * 1000
        stim_time = sb.df_epochs.at[e_idx, 'stimTime']
        offset_time = pre_time + stim_time
        # Convert from ms to samples
        onset_samples = int(np.round(pre_time * rb.amp_sample_rate / 1000))
        offset_samples = int(np.round(offset_time * rb.amp_sample_rate / 1000))
        sts = rb.spike_times[e_idx]
        n_sps = len(sts[(sts >= onset_samples) & (sts <= offset_samples)])
        if b_sub_baseline:
            n_baseline = len(sts[(sts < onset_samples)])
            n_sps -= n_baseline
        ls_nsps.append(n_sps)
    stim_spikes = np.array(ls_nsps)

    return stim_spikes

stim_spikes = get_stim_nspikes(sb, rb, b_sub_baseline=False)
```

```python
plt.scatter(spot_sizes, stim_spikes, c='k')
plt.xlabel('Spot size (um)')
plt.ylabel('Number of spikes during spot flash period')
```
